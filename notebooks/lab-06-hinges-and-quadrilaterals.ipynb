```python
def quadrilateral(t,sides):
    a,b,c,d=sides
    xb=(a*a+t*t-b*b)/(2*t)
    xd=(d*d+t*t-c*c)/(2*t)
    yb=np.sqrt(max(0.,a*a-xb*xb))
    yd=np.sqrt(max(0.,d*d-xd*xd))
    return np.array([[0.,0.],[xb,yb],[t,0.],[xd,-yd]])
```
